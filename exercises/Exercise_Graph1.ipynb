```python
from typing import Dict, TypedDict
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    name: str

def cheering_node(state: AgentState) -> AgentState:
    """A node that cheers the user on by modifying the 'name' field in the state."""
    
    state['name'] = state['name'] + ", you're doing an amazing job learning LangGraph!"

    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("cheering", cheering_node)

graph.set_entry_point("cheering")
graph.set_finish_point("cheering")

app = graph.compile()
```

```python
from IPython.display import display, Image
display(Image(app.get_graph().draw_mermaid_png()))
```

```python
result = app.invoke({"name": "Kübra"})
```
